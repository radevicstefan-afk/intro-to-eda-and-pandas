## Vectorized operations
---

Why bother with `DataFrame` and `Series` object when we could use Python lists and dictionaries?

`pandas` uses `numpy` arrays under the hood, which can perform operations **on an entire collection of values** rather than looping through the values individually.

Let's see a basic example.

We'll square the first 100,000 integers: first with a `for` loop, then as a `pandas` `Series` and compare timings.

*(we can use Jupyter's `%%timeit` command to time how long a cell takes to run)*

```python
salaries = range(1000000000)
```

Option 1: a `for` loop

```python
# Goal : Calculate how many villas people can buy with their salary

# What is the cost of one villa
villa_cost = 624890

# What are the salaries of the people? One billion people
numbers = range(1000000000)
```

```python
type(numbers)
```

```python
numbers
```

```python
# Step 1: divide the salary by the cost of one villa
example_salary = 100456
example_salary / villa_cost
```

```python
# Aleksandar answer
n_villas = example_salary // villa_cost
n_villas

# We mentioned % -> module
```

```python
import pandas as pd
salaries = range(20000000)
villa_cost = 634578
villas_to_buy = []
```

Option 1: a `for` loop

```python
# With the following line I calculate how much time this takes
%%timeit

# To calculate how many villas each person can buy
# I use a for loop and save every output in a list

for each_salary in salaries:
  # I calculate the division, round and save in the list in one line
  villas_to_buy.append(round(each_salary / villa_cost))
```

Option 2: `pandas` `Series`

```python
# With the following line I calculate how much time this takes
%%timeit

# Instead of the range, I will use a Series object from pandas
salaries_series = pd.Series(salaries)

# The final values are saved in another Series object
villas_to_buy = round(salaries_series / villa_cost)
```
